```python
import pandas as pd
from pathlib import Path
```

```python
mappings = Path('/mnt/raidbio2/extproj/projekte/textmining/mirnaTextmining/mirClassification/data/mirbase/mappings')
normalization = Path('/mnt/raidbio2/extproj/projekte/textmining/mirnaTextmining/mirClassification/data/mirbase/normalization')
families_path = mappings / 'families.tsv'
precursors_path = mappings / 'precursors.tsv'
mature_path = mappings / 'mature.tsv'
mipf_to_mi_path = mappings / 'mipf_to_mi_filtered.tsv'
mi_to_mimat_path = mappings/ 'mi_to_mimat.tsv'
family_normalization_path = normalization / 'family_normalization_dict.json'
precursor_normalization_path = normalization / 'precursor_normalization_dict.json'
mature_normalization_path = normalization / 'mature_normalization_dict.json'
```

```python
families = pd.read_csv(families_path, sep='\t')
precursors = pd.read_csv(precursors_path, sep='\t')
mature = pd.read_csv(mature_path, sep='\t')
mipf_tp_mi = pd.read_csv(mipf_to_mi_path, sep = '\t')
mi_to_mimat = pd.read_csv(mi_to_mimat_path, sep='\t')
```

```python
mapped_mipfs = set(mipf_tp_mi['MIPF'])
stored_mipfs = set(families['accession'])
mapped_mipfs - stored_mipfs
print(len(stored_mipfs - mapped_mipfs))
```

```python
mapped_mi = set(mipf_tp_mi['MI'])
stored_mi = set(precursors['accession'])
print(len(mapped_mi))
print(len(stored_mi))
print(len(mapped_mi - stored_mi))
print(len(stored_mi - mapped_mi)) # Not all of the precursors have a family
```

```python
mapped_mi_2 = set(mi_to_mimat['MI'])
print(mapped_mi_2 - stored_mi)
print(stored_mi - mapped_mi_2)
print(mapped_mi_2 == stored_mi)
```

```python
mapped_mimat = set(mi_to_mimat['MIMAT'])
stored_mimat = set(mature['accession'])
print(stored_mimat == mapped_mimat) # checks out
```

```python
import json
```

```python
with open(family_normalization_path, 'r') as fam:
    family_normalization_dict = json.load(fam)
with open(precursor_normalization_path, 'r') as pre:
    precursor_normalization_dict = json.load(pre)
with open(mature_normalization_path, 'r') as mat:
    mature_normalization_dict = json.load(mat)
```

```python
family_norm_ids = {acc for acc in family_normalization_dict.values()}
family_norm_ids == stored_mipfs # Check out
```

```python
precursor_norm_ids = {entry['accession'] for entry in precursor_normalization_dict.values() if not entry['dead']}
print(precursor_norm_ids - stored_mi)
stored_mi - precursor_norm_ids
```

```python
mature_norm_ids = {entry['accession'] for entry in mature_normalization_dict.values() if not entry['dead']}
print(mature_norm_ids - stored_mimat)
print(len(stored_mimat - mature_norm_ids))
```
