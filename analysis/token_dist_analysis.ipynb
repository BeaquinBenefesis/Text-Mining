```python
import pandas as pd
from textmining.paths import OUTPUTS_DIR
```

```python
df_hmdd = pd.read_csv(OUTPUTS_DIR / 'HMDD' / 'distances.txt' , sep='\t')
df_bg = pd.read_csv(OUTPUTS_DIR / 'disease_and_mir' / 'distances.txt', sep='\t')
```

```python
import matplotlib.pyplot as plt

series = df_hmdd['distance']

series.hist(bins=50)
plt.xlabel('token distance')
plt.ylabel('count')
plt.title('Distance distribution')
plt.show()
```

```python
hmdd_series = df_hmdd['distance']
bg_sample = df_bg['distance'].sample(n=len(hmdd_series), random_state=42)

max_val = max(hmdd_series.max(), bg_sample.max())
bins = range(0, max_val + 2)

plt.hist(bg_sample, bins=bins, alpha=0.5, label='background')
plt.hist(hmdd_series, bins=bins, alpha=0.5, label='HMDD')
plt.xlabel('token distance')
plt.ylabel('count')
plt.title('Distance distribution (overlapped)')
plt.legend()
plt.show()
```
